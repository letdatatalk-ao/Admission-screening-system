```python
import pytesseract
from pdf2image import convert_from_path

pytesseract.pytesseract.tesseract_cmd = r"C:\Program Files\Tesseract-OCR\tesseract.exe"

pages = convert_from_path(
    "../data/raw/transcripts/225127909 T.pdf",
    dpi=300,
    poppler_path=r"C:\Users\User\Downloads\Release-25.12.0-0\poppler-25.12.0\Library\bin"
)

# Score de confiance page par page
for i, page_img in enumerate(pages):
    data = pytesseract.image_to_data(
        page_img,
        lang="eng",
        output_type=pytesseract.Output.DICT
    )
    confidences = [c for c in data["conf"] if c != -1]
    avg_conf = sum(confidences) / len(confidences) if confidences else 0
    print(f"Page {i+1} — score OCR moyen : {avg_conf:.1f}/100")

    if avg_conf < 40:
        print("  → POOR_QUALITY flag : needs_human_review = True")
    else:
        print("  → qualité acceptable")
```
